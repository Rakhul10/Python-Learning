```python
x = b"Hello"
print(x)
print(type(x))
#It tells Python to store the string as a sequence of bytes 
# (specifically, ASCII/UTF-8 byte values) rather than a standard Unicode str object. 
# When you check its data type using the type() function, it returns bytes.
```

```python
x = bytearray(5)
print(x)
print(type(x))
x[0] = 65  # Modifies the first byte to ASCII 'A'
print(x)   # Output: bytearray(b'A\x00\x00\x00\x00')
```

```python
x = memoryview(bytes(5))
print(x)
#Creates a memory view object referencing the buffer of bytes(5)
#A memoryview allows Python code to access the internal data of an object (that supports the buffer protocol, like bytes or bytearray) directly without making a copy of the data first.
#  This is primarily used for high-performance memory management and low-level data processing (e.g., handling large binary files or socket data).
```
