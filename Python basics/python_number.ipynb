python Number
There are three numeric types in Python:

int
float
complex


```python
#float
x = 35e3
y = 12E4
z = -87.7e100

print(type(x))
print(type(y))
print(type(z))
```

```python
#complex
x = 3+5j
y = 5j
z = -5j

print(type(x))
print(type(y))
print(type(z))
```

```python
#TYPE CONVERSION
x = 1    # int
y = 2.8  # float
z = 1j   # complex

#convert from int to float:
a = float(x)

#convert from float to int:
b = int(y)

#convert from int to complex:
c = complex(x)

print(a)
print(b)
print(c)

print(type(a))
print(type(b))
print(type(c))
```

```python
x = 15
y = 4

print(x + y)
print(x - y)
print(x * y)
print(x / y)
print(x % y)
print(x ** y)
print(x // y)
```

```python
print(x := 3)
```

```python
#The Ternary Operator
#allows you to assign one value if a condition is true, and another if it is false
num = 6

x = "WEEKEND!" if num > 5 else "Workday"

print(x)


x = "Fri" if num == 5 else "Sat" if num == 6 else "Sun" if num == 7 else "weekday"

print(x)
```

```python
#bitwise operators
print(6 & 3)  # Output: 2 -> 6in binary is 110, and 3 in binary is 011. The bitwise AND operation results in 010, which is 2 in decimal.
print(6 | 3)  # Output: 7 -> 6in binary is 110, and 3 in binary is 011. The bitwise OR operation results in 111, which is 7 in decimal.
print(6 ^ 3)  # Output: 5 -> 6in binary is 110, and 3 in binary is 011. The bitwise XOR operation results in 101, which is 5 in decimal.
```
